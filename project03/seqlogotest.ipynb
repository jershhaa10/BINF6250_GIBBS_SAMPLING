```python
import seqlogo as sl
import pandas as pd
import numpy as np

from working_test import GibbsMotifSampler, get_reads_from_bam

seed=123

seqs = get_reads_from_bam("data/SRR9090854.subsampled_5pct.bam",100,seed=seed)

pfm = GibbsMotifSampler(
    seqs=seqs, 
    k = 9, seed = seed,
    bg_pseudocount = 1.0, 
    motif_pseudocount = 0.25,
    max_tries_init = 10, 
    burn_in_sweeps = 1,
    check_every_sweeps = 1,
    rel_thresh = 0.01,
    max_sweeps = 10
    )

sl.Pfm(pfm)

sl.seqlogo(sl.CompletePm(pfm = pfm.T))
```

```python
# this is the documentation example
random_ppm = np.random.dirichlet(np.ones(4), size=6)
ppm = sl.Ppm(random_ppm)
sl.seqlogo(ppm, ic_scale = False, format = 'png', size = 'medium')
```
